## Score A Protein Against Explicit SMILES

```python
from omtx.lula import load_model\n\nprotein_sequence = "MEEPQSDPSV"\nsmiles = [\n    "CCO",\n    "c1ccccc1",\n    "CC(=O)Nc1nnc(s1)S(N)(=O)=O",\n]\n\nmodel = load_model(MODEL_ID)\nrows = sorted(\n    model.score(protein_sequence=protein_sequence, smiles=smiles),\n    key=lambda row: row["rank"],\n)\nrows
```
